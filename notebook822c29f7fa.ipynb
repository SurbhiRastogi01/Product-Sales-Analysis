```python
# Convert OrderDate and DeliveryDate to datetime

df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce')
df['DeliveryDate'] = pd.to_datetime(df['DeliveryDate'], errors='coerce')

print("OrderDate type:", df['OrderDate'].dtype)
print("DeliveryDate type:", df['DeliveryDate'].dtype)
```

```python
# Calculate delivery time in days

df['DeliveryDays'] = (
    df['DeliveryDate'] - df['OrderDate']
).dt.days

display(df[['OrderDate', 'DeliveryDate', 'DeliveryDays']].head(10))
```

```python
# Delivery time statistics

print("Average Delivery Days:", df['DeliveryDays'].mean())
print("Minimum Delivery Days:", df['DeliveryDays'].min())
print("Maximum Delivery Days:", df['DeliveryDays'].max())

display(df['DeliveryDays'].describe())
```

```python
# Overall Sales Analysis

total_sales = df['TotalPrice'].sum()
average_sales = df['TotalPrice'].mean()
maximum_sale = df['TotalPrice'].max()
minimum_sale = df['TotalPrice'].min()

print("Total Sales:", total_sales)
print("Average Order Value:", average_sales)
print("Maximum Order Value:", maximum_sale)
print("Minimum Order Value:", minimum_sale)
```

```python
# Sales by Region

region_sales = (
    df.groupby('Region')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
)

display(region_sales)
```

```python
import matplotlib.pyplot as plt

# Calculate sales by region
region_sales = df.groupby('Region')['TotalPrice'].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(10, 5))
region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 2: Sales by Salesperson

salesperson_sales = (
    df.groupby('Salesperson')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
salesperson_sales.plot(kind='bar')

plt.title('Total Sales by Salesperson')
plt.xlabel('Salesperson')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Top 10 Salespeople by Total Sales

top_salespeople = (
    df.groupby('Salesperson')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
top_salespeople.plot(kind='bar')

plt.title('Top 10 Salespeople by Total Sales')
plt.xlabel('Salesperson')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 3: Returned vs Non-Returned Orders

return_counts = df['Returned'].value_counts()

plt.figure(figsize=(7, 5))
return_counts.plot(kind='bar')

plt.title('Returned vs Non-Returned Orders')
plt.xlabel('Returned (0 = No, 1 = Yes)')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Graph 4: Sales by Payment Method

payment_sales = (
    df.groupby('PaymentMethod')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
payment_sales.plot(kind='bar')

plt.title('Total Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 5: Sales by Promotion

promotion_sales = (
    df.groupby('Promotion')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
promotion_sales.plot(kind='bar')

plt.title('Total Sales by Promotion')
plt.xlabel('Promotion')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 6: Monthly Sales Trend

# Create Month column
df['Month'] = df['OrderDate'].dt.to_period('M')

# Calculate monthly sales
monthly_sales = (
    df.groupby('Month')['TotalPrice']
      .sum()
)

# Convert period to string for plotting
monthly_sales.index = monthly_sales.index.astype(str)

# Create line chart
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Graph 7: Top 10 Products by Quantity Sold

product_quantity = (
    df.groupby('Product')['Quantity']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
product_quantity.plot(kind='bar')

plt.title('Top 10 Products by Quantity Sold')
plt.xlabel('Product')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 8: Average Delivery Days by Region

delivery_by_region = (
    df.groupby('Region')['DeliveryDays']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
delivery_by_region.plot(kind='bar')

plt.title('Average Delivery Days by Region')
plt.xlabel('Region')
plt.ylabel('Average Delivery Days')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Graph 9: Discount vs Total Sales

plt.figure(figsize=(10, 6))

plt.scatter(df['Discount'], df['TotalPrice'], alpha=0.5)

plt.title('Discount vs Total Sales')
plt.xlabel('Discount')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
# Graph 10: Sales vs Quantity

plt.figure(figsize=(10, 6))

plt.scatter(df['Quantity'], df['TotalPrice'], alpha=0.5)

plt.title('Sales vs Quantity')
plt.xlabel('Quantity Sold')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
# Graph 11: Top 10 Products by Total Sales

product_sales = (
    df.groupby('Product')['TotalPrice']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
product_sales.plot(kind='bar')

plt.title('Top 10 Products by Total Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Graph 12: Return Rate by Region

return_rate = (
    df.groupby('Region')['Returned']
      .mean() * 100
)

return_rate = return_rate.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
return_rate.plot(kind='bar')

plt.title('Return Rate by Region')
plt.xlabel('Region')
plt.ylabel('Return Rate (%)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Graph 13: Average Shipping Cost by Region

shipping_by_region = (
    df.groupby('Region')['ShippingCost']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
shipping_by_region.plot(kind='bar')

plt.title('Average Shipping Cost by Region')
plt.xlabel('Region')
plt.ylabel('Average Shipping Cost')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Graph 14: Correlation Heatmap

import seaborn as sns
import matplotlib.pyplot as plt

numeric_data = df.select_dtypes(include=['int64', 'float64'])

correlation = numeric_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

```python
# Step 15: Key Business Insights

print("========== KEY BUSINESS INSIGHTS ==========")

# 1. Best performing region
best_region = df.groupby('Region')['TotalPrice'].sum().idxmax()
best_region_sales = df.groupby('Region')['TotalPrice'].sum().max()

print(f"\n1. Highest Sales Region: {best_region}")
print(f"   Total Sales: {best_region_sales:.2f}")

# 2. Best performing product
best_product = df.groupby('Product')['TotalPrice'].sum().idxmax()
best_product_sales = df.groupby('Product')['TotalPrice'].sum().max()

print(f"\n2. Highest Selling Product: {best_product}")
print(f"   Total Sales: {best_product_sales:.2f}")

# 3. Best salesperson
best_salesperson = df.groupby('Salesperson')['TotalPrice'].sum().idxmax()
best_salesperson_sales = df.groupby('Salesperson')['TotalPrice'].sum().max()

print(f"\n3. Highest Salesperson: {best_salesperson}")
print(f"   Total Sales: {best_salesperson_sales:.2f}")

# 4. Average delivery time
print(f"\n4. Average Delivery Time: {df['DeliveryDays'].mean():.2f} days")

# 5. Return rate
return_rate = df['Returned'].mean() * 100
print(f"\n5. Overall Return Rate: {return_rate:.2f}%")

# 6. Average discount
print(f"\n6. Average Discount: {df['Discount'].mean():.2f}")

# 7. Average shipping cost
print(f"\n7. Average Shipping Cost: {df['ShippingCost'].mean():.2f}")

print("\n==========================================")
```
